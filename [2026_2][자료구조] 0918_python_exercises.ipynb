### list 사용법
- `[]` 로 만들고, 순서가 있어서 인덱스(0부터)로 접근합니다.
- `append()` 로 맨 뒤에 값을 추가합니다.
- 값이 중복돼도 그대로 다 저장됩니다.

````python
fruits = ["apple", "banana", "cherry"]

# 전체 출력
print(fruits)

# 인덱스로 접근 (첫 번째 값)
print(fruits[0])

# 개수 확인
print(len(fruits))

# 맨 뒤에 추가
fruits.append("mango")
print(fruits)

# 중간에 추가
fruits.insert(2, "orange")
print(fruits)

# for문으로 하나씩 순회
for i in fruits:
    print(i)
````

### set 사용법

````python
numbers = {1, 2, 2, 3, 3, 3}

# 중복이 자동으로 사라짐: {1, 2, 3}
print(numbers)            

# 값 추가
numbers.add(4)
print(numbers)            

# 포함 여부 확인 -> True
print(5 in numbers)

# -> False
   

my_list = [1, 2, 2, 3, 3, 3]

# 리스트를 set으로 변환하여 출력해보세요 (차이??)
my_set = set(my_list)
print(my_set)

# 다시 리스트로 되돌리기
my_list = list(my_set)
print(my_list)
````

### dict 사용법
- `{key: value}` 형태로 만들고, 순서가 아니라 **키(key)로 값을 찾습니다.
- 값을 꺼낼 땐 `dict[키]`, 없는 키일 수도 있으면 `dict.get(키)`가 더 안전합니다.
- 키가 이미 있으면 값을 덮어쓰고, 없으면 새로 추가됩니다.

````python
scores = {"james":80, "jenny": 90}

# 전체 출력
print(scores)

# 키로 값 꺼내기
print(scores["james"])

# 없는 키 -> None (에러 안 남)


# 이렇게 하면 KeyError 발생 (주석 풀어서 확인해보세요)
# print(scores["aiden"])     

# 새 키 추가
scores["jinny"] = 100

# 기존 키는 값만 덮어씀
scores["jenny"] = 95

# 키만 모아서 보기
print(scores.keys())

# 값만 모아서 보기
print(scores.values())

# (키, 값) 쌍으로 보기
print(scores.items())

# 키, 값 동시에 순회
for k, v in scores.items():
    print(k, ":", v)


# 키가 존재하는지 확인 -> True
print("jenny" in scores.keys())

# 키 삭제
del scores["jinny"]
print(scores)

# new_key 변수에 andy를 저장하세요
new_key = "jinny"

# new_value 변수에 점수를 저장하세요
new_value = 100

# dict scores 에 저장하세요
scores[new_key] = new_value
print(scores)
````

### Q1. 짝수만 골라내기
정수 리스트를 받아서 짝수만 골라 새 리스트로 반환하는 함수를 작성하세요.

````python
def get_even_numbers(nums):
    result = []
    for i in nums:
        if i % 2 == 0:
            result.append(i)
    return result
````

### Q2. 최댓값 찾기 (직접 구현)
`max()` 내장 함수를 쓰지 말고 반복문으로 직접 비교해서 가장 큰 값을 찾으세요.

````python
def find_max(nums):
    biggest = nums[0]
    for i in nums:
        if biggest < i:
            biggest = i
    return biggest
````

````python
# 여기에 함수를 직접 호출해서 결과를 확인해보세요
my_list = [1, 2, 3, 4, 5]
new_list = find_max(my_list)
print(new_list)
````

### Q4. 평균 점수 구하기
학생 이름을 key, 점수를 value로 갖는 dict를 받아 전체 평균 점수를 반환하세요.
힌트: `scores.values()` 로 값만 뽑을 수 있습니다.

````python
def get_average_score(scores: dict):
    total = 0
    count = 0
    for v in scores.values():
        total += v
        count += 1
    return total / count if count > 0 else 0
````

````python
# 여기에 함수를 직접 호출해서 결과를 확인해보세요
st_dict = {
    "james" : 60,
    "jenny" : 80,
    "jinny" : 100,
}
print(get_average_score(st_dict))
````

````python
def add_stock(inventory: dict, item: str, count: int):
    if item in inventory.keys():
        inventory.values += count
    else:
        inventory[item] = count
    pass


def print_inventory(inventory: dict):
    for k, v in inventory.items():
        print(k,": ", v)
    pass
````

## Part 3. 배열(Array)과 해시(Hash) 
파이썬에서는 `list`를 '배열'처럼, `dict`/`set`을 '해시테이블'처럼 사용합니다.

### 해시(Hash) 사용법
- 배열(list)에서 값을 찾으려면 앞에서부터 하나씩 확인해야 합니다. (O(n))
- dict는 "해시 함수"로 key가 저장된 위치를 바로 계산해서 찾기 때문에, 데이터가 아무리 많아져도 거의 즉시 찾습니다. (평균 O(1))
- 그래서 "이름으로 전화번호 찾기"처럼 key로 값을 조회하는 상황에서는 list보다 dict(해시테이블)를 씁니다.

````python
phonebook = {"cassie": "010-1111", "brenda": "010-2222"}

# 있는 키를 조회하는 두 가지 방법
print(phonebook["cassie"])
print(phonebook.get("brenda"))

# 없는 키를 조회하면?
      # KeyError 발생! 주석 풀어서 직접 확인해보세요
      # 에러 없이 None만 반환 (더 안전함)

# in 으로 키가 있는지 먼저 확인하기
print("cassie" in phonebook)

# 비교: 만약 리스트였다면 이렇게 하나씩 다 뒤져야 합니다 (O(n))
people = [("cassie", "010-1111"), ("brenda", "010-2222")]
for name, number in people:
      print(name, ": ", number)
````

### Q6. [배열] 순차 탐색
배열(리스트)을 앞에서부터 하나씩 확인해서 target이 몇 번째 칸(인덱스)에 있는지 반환하세요. 없으면 -1.
(강의자료 66쪽 그림과 같은 방식, O(n))

````python
def linear_search_index(arr, target):
    for i in range(len(arr)):
        if i == target:
            return arr.index(target)
        pass
    return -1
````

````python
# 여기에 함수를 직접 호출해서 결과를 확인해보세요

arr = [3, 7, 2, 9, 4]
linear_search_index(arr, 3)
````

### heapq 사용법
- 파이썬은 리스트를 '최소 힙(min-heap)'으로 다룰 수 있게 해주는 `heapq` 모듈을 기본 제공합니다.
- 힙은 **가장 작은 값을 빠르게 꺼내는 것**에 최적화된 구조입니다. (일반 리스트에서 최솟값을 찾으려면 O(n)이지만, 힙은 O(log n))

````python
import heapq

nums = [5, 1, 8, 3, 9, 2]

# nums 자체를 힙 구조로 변환 (완전히 정렬된 건 아님)
heapq.heapify(nums)
print(nums)

# nums[0]은 항상 최솟값
````

````python
# 값 삽입

heapq.heappush(nums, 100)
print(nums)
````

````python
# 가장 작은 값 꺼내기

heapq.heappushpop(nums, nums[0])
````

### Q9. STEP 카운터로 실제 연산 횟수 세어보기
반복문이 몇 번 실행되는지 직접 세어보는 함수입니다. n을 10, 100, 1000으로 바꿔가며 실행해서 STEP이 어떻게 변하는지 관찰하세요.

````python
def sum_with_step_counter(n):
    total = 0
    step = 0
    for i in range(1, n + 1):
        total = total + i
        step += 1  # 이 줄이 몇 번 실행되는지가 곧 시간복잡도
    return total, step
````

````python
# 여기에 함수를 직접 호출해서 결과를 확인해보세요

print(sum_with_step_counter(100))
print(sum_with_step_counter(500))
print(sum_with_step_counter(2500))
````

### Q10. Big-O 순서 체감해보기

아래 네 가지 시간복잡도를 수행시간이 오래 걸리는 순서대로 나열해봅시다.

- O(7n log n + 100)
- O(3ⁿ⁺²)
- O(n¹⁰)
- O(5 · n!)

머릿속으로 순서를 예상해본 뒤, 아래 코드로 실제 연산 횟수와 예상 소요시간을 계산해서 확인해보세요.
(컴퓨터가 1초에 대략 1억(10⁸)번 연산한다고 가정하고 초 단위로 환산합니다.)


````python
import math

def op_count_nlogn(n):
    return 7 * n * math.log2(n) + 100


def op_count_exp(n):
    return 3 ** (n + 2)


def op_count_poly(n):
    return n ** 10


def op_count_factorial(n):
    return 5 * math.factorial(n)
````

````python
OPS_PER_SECOND = 10 ** 8  # 컴퓨터가 1초에 대략 이만큼 연산한다고 가정

def format_seconds(sec):
    if sec < 1:
        return f"{sec:.6f}초"
    elif sec < 60:
        return f"{sec:.2f}초"
    elif sec < 3600:
        return f"{sec/60:.2f}분"
    elif sec < 86400:
        return f"{sec/3600:.2f}시간"
    elif sec < 31_536_000:
        return f"{sec/86400:.2f}일"
    else:
        return f"{sec/31_536_000:.2e}년"


def compare_big_o(n):
    formulas = {
        "O(7n log n + 100)": op_count_nlogn(n),
        "O(3^(n+2))": op_count_exp(n),
        "O(n^10)": op_count_poly(n),
        "O(5 * n!)": op_count_factorial(n),
    }
    print(f"--- n = {n} ---")
    for name, count in sorted(formulas.items(), key=lambda x: x[1]):
        seconds = count / OPS_PER_SECOND
        print(f"{name:20s} 연산 횟수 ≈ {count:.3e}   예상 소요시간 ≈ {format_seconds(seconds)}")
````

````python
# 여기에 함수를 직접 호출해서 결과를 확인해보세요
compare_big_o(5)
compare_big_o(10)
compare_big_o(100)

````
